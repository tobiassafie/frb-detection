# FRB CNN Classifier — Training Notebook

**Fallback from LSTM** — LSTM won't synthesize at 256 timesteps.

Uses the same v6 dataset `(256, 256)` reshaped to `(256, 256, 1)` for Conv2D.

Architecture based on the pooled CNN that already synthesized in <2 minutes on the VU9P.

**FPGA constraints:**
- Flatten dimension < 500
- < 50k parameters
- io_stream + Resource strategy

```python
import numpy as np
import matplotlib.pyplot as plt
import os
import json
from datetime import datetime

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, optimizers
from sklearn.metrics import (
    confusion_matrix, classification_report,
    roc_curve, auc, precision_recall_curve
)

print(f"TensorFlow: {tf.__version__}")
print(f"GPUs: {tf.config.list_physical_devices('GPU')}")

set_seed = 42
tf.random.set_seed(set_seed)
np.random.seed(set_seed)
```

```python
# ============================================================
# ALL CONFIGURATION IN ONE PLACE
# ============================================================
CONFIG = {
    # --- Data ---
    "data_path": "./augmented_dataset/lstm_dataset_v6.npz",
    
    # --- CNN Architecture ---
    # Each entry: (filters, kernel_size)
    # MaxPool2D(2,2) is applied after each conv layer
    "conv_layers": [
        (2, 3),     # 256→128
        (4, 3),     # 128→64
        (4, 3),     # 64→32
        (4, 3),     # 32→16
        (4, 3),     # 16→8
    ],
    "conv_activation": "relu",
    "pool_size": (2, 2),
    
    "dense_units": [32],       # Dense layers after flatten
    "dense_dropout": 0.3,
    "dense_activation": "relu",
    
    "output_classes": 1,       # 1 = binary sigmoid, 2 = softmax
    
    # --- Training ---
    "batch_size": 128,
    "epochs": 150,
    "learning_rate": 1e-3,
    "optimizer": "adam",
    "lr_schedule": "reduce_on_plateau",
    "lr_patience": 10,
    "lr_factor": 0.5,
    "early_stop_patience": 30,
    "class_weight": "auto",
    
    # --- Output ---
    "save_dir": "./models/",
    "experiment_name": "cnn_v1",
}

print("Configuration:")
for k, v in CONFIG.items():
    print(f"  {k}: {v}")
```

## 1. Load Data

```python
data = np.load(CONFIG["data_path"])
X_train, y_train = data["X_train"], data["y_train"]
X_val, y_val = data["X_val"], data["y_val"]
X_test, y_test = data["X_test"], data["y_test"]

print(f"Loaded: {X_train.shape}")

# Reshape (batch, 256, 256) → (batch, 256, 256, 1) for Conv2D
X_train = X_train[..., np.newaxis]
X_val = X_val[..., np.newaxis]
X_test = X_test[..., np.newaxis]

input_shape = X_train.shape[1:]  # (256, 256, 1)

print(f"Reshaped for CNN: {X_train.shape}")
print(f"Input shape: {input_shape}")
print(f"Train: {X_train.shape[0]} ({np.mean(y_train):.1%} FRB)")
print(f"Val:   {X_val.shape[0]} ({np.mean(y_val):.1%} FRB)")
print(f"Test:  {X_test.shape[0]} ({np.mean(y_test):.1%} FRB)")
```

## 2. Build Model

Key: MaxPooling aggressively reduces spatial dimensions so the flatten is small.

```python
def build_cnn(config, input_shape):
    model = keras.Sequential()
    model.add(layers.Input(shape=input_shape, name='input_data'))
    
    # Conv + Pool blocks
    for i, (filters, kernel) in enumerate(config["conv_layers"]):
        model.add(layers.Conv2D(
            filters, (kernel, kernel),
            activation=config["conv_activation"],
            padding='same',
            name=f'conv_{i}'
        ))
        model.add(layers.MaxPooling2D(
            pool_size=config["pool_size"],
            name=f'pool_{i}'
        ))
    
    # Flatten
    model.add(layers.Flatten(name='flatten'))
    
    # Dense head
    for j, units in enumerate(config["dense_units"]):
        model.add(layers.Dense(
            units, activation=config["dense_activation"],
            name=f'dense_{j}'
        ))
        if config["dense_dropout"] > 0:
            model.add(layers.Dropout(config["dense_dropout"], name=f'drop_{j}'))
    
    # Output
    if config["output_classes"] == 1:
        model.add(layers.Dense(1, activation='sigmoid', name='output'))
    else:
        model.add(layers.Dense(config["output_classes"], activation='softmax', name='output'))
    
    return model

model = build_cnn(CONFIG, input_shape)
model.summary()

# Check flatten dimension — compute from last pooling layer output
# After 5 pool layers: 256 / 2^5 = 8, so 8 x 8 x last_filters
n_pools = len(CONFIG["conv_layers"])
spatial = 256 // (2 ** n_pools)
last_filters = CONFIG["conv_layers"][-1][0]
flatten_dim = spatial * spatial * last_filters

print(f"\nFlatten dimension: {flatten_dim}")
print(f"  ({spatial} x {spatial} x {last_filters})")
if flatten_dim > 500:
    print(f"⚠️  Over 500 — add more pooling or reduce filters")
else:
    print(f"✅ Under 500 — synthesis-friendly")

total_params = model.count_params()
print(f"\nTotal parameters: {total_params:,}")
print(f"{'✅ Under 50k' if total_params < 50000 else '⚠️  Over 50k FPGA budget'}")
```

## 3. Compile & Train

```python
# Optimizer
opt_map = {
    "adam": optimizers.Adam(learning_rate=CONFIG["learning_rate"]),
    "adamw": optimizers.AdamW(learning_rate=CONFIG["learning_rate"]),
    "sgd": optimizers.SGD(learning_rate=CONFIG["learning_rate"], momentum=0.9),
}
optimizer = opt_map[CONFIG["optimizer"]]

loss = 'binary_crossentropy' if CONFIG["output_classes"] == 1 else 'sparse_categorical_crossentropy'

model.compile(
    optimizer=optimizer,
    loss=loss,
    metrics=['accuracy'],
)

# Class weights
if CONFIG["class_weight"] == "auto":
    n_pos = np.sum(y_train == 1)
    n_neg = np.sum(y_train == 0)
    class_weight = {0: len(y_train) / (2 * n_neg), 1: len(y_train) / (2 * n_pos)}
    print(f"Class weights: {class_weight}")
elif CONFIG["class_weight"] == "none":
    class_weight = None
else:
    class_weight = CONFIG["class_weight"]

# Callbacks
save_dir = os.path.join(CONFIG["save_dir"], CONFIG["experiment_name"])
os.makedirs(save_dir, exist_ok=True)

cb_list = [
    callbacks.EarlyStopping(
        monitor='val_loss', patience=CONFIG['early_stop_patience'],
        restore_best_weights=True, verbose=1
    ),
    callbacks.ModelCheckpoint(
        os.path.join(save_dir, 'best_model.keras'),
        monitor='val_loss', save_best_only=True, verbose=0
    ),
]

if CONFIG['lr_schedule'] == 'reduce_on_plateau':
    cb_list.append(callbacks.ReduceLROnPlateau(
        monitor='val_loss', factor=CONFIG['lr_factor'],
        patience=CONFIG['lr_patience'], min_lr=1e-6, verbose=1
    ))

print(f"Callbacks: {[type(cb).__name__ for cb in cb_list]}")
```

```python
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=CONFIG['epochs'],
    batch_size=CONFIG['batch_size'],
    class_weight=class_weight,
    callbacks=cb_list,
    verbose=1,
)
```

```python
# Training curves
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(history.history['loss'], label='Train', linewidth=2)
axes[0].plot(history.history['val_loss'], label='Val', linewidth=2)
axes[0].set_title('Loss')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Binary Crossentropy')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(history.history['accuracy'], label='Train', linewidth=2)
axes[1].plot(history.history['val_accuracy'], label='Val', linewidth=2)
axes[1].set_title('Accuracy')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.suptitle(f"{CONFIG['experiment_name']} — Training Curves", fontsize=13)
plt.tight_layout()
plt.savefig(os.path.join(save_dir, 'training_curves.png'), dpi=150)
plt.show()

best_epoch = np.argmin(history.history['val_loss'])
print(f"Best epoch: {best_epoch + 1}")
print(f"  Val loss: {history.history['val_loss'][best_epoch]:.4f}")
print(f"  Val acc:  {history.history['val_accuracy'][best_epoch]:.4f}")
```

## 4. Evaluate

```python
y_pred_prob = model.predict(X_test, batch_size=CONFIG['batch_size']).flatten()
y_pred = (y_pred_prob >= 0.5).astype(int)

test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Test Loss:     {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
print(f"\n{classification_report(y_test, y_pred, target_names=['Noise', 'FRB'])}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
im = axes[0].imshow(cm, cmap='Blues')
axes[0].set_xticks([0, 1]); axes[0].set_yticks([0, 1])
axes[0].set_xticklabels(['Noise', 'FRB'])
axes[0].set_yticklabels(['Noise', 'FRB'])
axes[0].set_xlabel('Predicted'); axes[0].set_ylabel('True')
axes[0].set_title('Confusion Matrix')
for i in range(2):
    for j in range(2):
        axes[0].text(j, i, f'{cm[i,j]}', ha='center', va='center', fontsize=16,
                     color='white' if cm[i,j] > cm.max()/2 else 'black')

# ROC
fpr, tpr, _ = roc_curve(y_test, y_pred_prob)
roc_auc = auc(fpr, tpr)
axes[1].plot(fpr, tpr, linewidth=2, label=f'AUC = {roc_auc:.4f}')
axes[1].plot([0,1], [0,1], 'k--', alpha=0.3)
axes[1].set_xlabel('FPR'); axes[1].set_ylabel('TPR')
axes[1].set_title('ROC Curve'); axes[1].legend(); axes[1].grid(True, alpha=0.3)

# PR
precision, recall, _ = precision_recall_curve(y_test, y_pred_prob)
pr_auc = auc(recall, precision)
axes[2].plot(recall, precision, linewidth=2, label=f'PR AUC = {pr_auc:.4f}')
axes[2].set_xlabel('Recall'); axes[2].set_ylabel('Precision')
axes[2].set_title('Precision-Recall'); axes[2].legend(); axes[2].grid(True, alpha=0.3)

plt.suptitle(f"{CONFIG['experiment_name']} — Test Evaluation", fontsize=13)
plt.tight_layout()
plt.savefig(os.path.join(save_dir, 'evaluation.png'), dpi=150)
plt.show()
```

```python

def progressive_probability(model, patch, n_steps=64, fill_value=None):
    """
    Reveal time bins left-to-right, predict P(FRB) at each step.
    
    fill_value: what to put in unrevealed bins.
        None  → use the patch's own mean (safest, no texture boundary)
        0.0   → zeros
        'noise' → random normal matching patch stats
    """
    n_time = patch.shape[0]  # 256
    step_size = max(1, n_time // n_steps)
    
    probs = []
    time_fracs = []
    
    for t in range(step_size, n_time + 1, step_size):
        masked = np.copy(patch)
        
        if fill_value is None:
            # Fill with mean of the revealed portion
            masked[t:, :] = masked[:t, :].mean()
        elif fill_value == 'noise':
            # Fill with Gaussian noise matching revealed portion stats
            revealed = masked[:t, :]
            masked[t:, :] = np.random.normal(
                revealed.mean(), max(revealed.std(), 1e-6), masked[t:, :].shape
            )
        else:
            masked[t:, :] = fill_value
        
        # Predict — model expects (batch, 256, 256, 1)
        pred = model.predict(masked[np.newaxis, ..., np.newaxis], verbose=0)[0, 0]
        probs.append(pred)
        time_fracs.append(t / n_time)
    
    return np.array(time_fracs), np.array(probs)


# --- Run on a few FRB and noise test examples ---
rng = np.random.default_rng(42)

n_examples = 5
frb_idxs = rng.choice(np.where(y_test == 1)[0], n_examples, replace=False)
noise_idxs = rng.choice(np.where(y_test == 0)[0], n_examples, replace=False)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# --- Panel 1: FRB examples, fill=mean ---
ax = axes[0]
for i, idx in enumerate(frb_idxs):
    patch = X_test[idx, :, :, 0]  # strip channel dim
    frac, prob = progressive_probability(model, patch, n_steps=64, fill_value=None)
    ax.plot(frac, prob, alpha=0.7, label=f'FRB #{i}')
ax.axhline(0.5, color='r', ls='--', alpha=0.5, label='threshold')
ax.set_xlabel('Fraction of time bins revealed')
ax.set_ylabel('P(FRB)')
ax.set_title('FRB patches (fill = mean)')
ax.set_ylim(-0.05, 1.05)
ax.legend(fontsize=8)
ax.grid(True, alpha=0.3)

# --- Panel 2: Noise examples, fill=mean ---
ax = axes[1]
for i, idx in enumerate(noise_idxs):
    patch = X_test[idx, :, :, 0]
    frac, prob = progressive_probability(model, patch, n_steps=64, fill_value=None)
    ax.plot(frac, prob, alpha=0.7, label=f'Noise #{i}')
ax.axhline(0.5, color='r', ls='--', alpha=0.5, label='threshold')
ax.set_xlabel('Fraction of time bins revealed')
ax.set_ylabel('P(FRB)')
ax.set_title('Noise patches (fill = mean)')
ax.set_ylim(-0.05, 1.05)
ax.legend(fontsize=8)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('figures/progressive_probability.svg')
plt.show()

# --- Panel 3: Diagnose fill value effect on ONE FRB ---
'''
ax = axes[2]
patch = X_test[frb_idxs[0], :, :, 0]

for fill, ls, lbl in [(None, '-', 'fill=mean'), (0.0, '--', 'fill=0'), ('noise', ':', 'fill=noise')]:
    frac, prob = progressive_probability(model, patch, n_steps=64, fill_value=fill)
    ax.plot(frac, prob, ls=ls, linewidth=2, label=lbl)

ax.axhline(0.5, color='r', ls='--', alpha=0.5)
ax.set_xlabel('Fraction of time bins revealed')
ax.set_ylabel('P(FRB)')
ax.set_title('Fill value comparison (same FRB patch)')
ax.set_ylim(-0.05, 1.05)
ax.legend(fontsize=8)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
'''

# --- Quick summary stats ---
print("\nBaseline P(FRB) at 1/64 revealed (≈4 time bins):")
for fill_name, fill_val in [("mean", None), ("zero", 0.0), ("noise", "noise")]:
    probs_at_start = []
    for idx in np.concatenate([frb_idxs, noise_idxs]):
        patch = X_test[idx, :, :, 0]
        _, prob = progressive_probability(model, patch, n_steps=64, fill_value=fill_val)
        probs_at_start.append(prob[0])
    print(f"  fill={fill_name:5s}:  mean={np.mean(probs_at_start):.3f}  "
          f"min={np.min(probs_at_start):.3f}  max={np.max(probs_at_start):.3f}")
```

```python
def find_burst_onset(patch, threshold_sigma=2.5, freq_frac=0.1):
    """
    Find the first time bin where the burst signal appears.
    
    Looks at only the TOP freq_frac of frequency channels (highest freq),
    where the dispersive sweep arrives first. This avoids diluting the 
    signal across 256 channels when the sweep only occupies a few at 
    any given time bin.
    
    Lower threshold_sigma (2.5 vs 3.0) to catch weaker injections.
    """
    n_time, n_freq = patch.shape
    
    # Top fraction of frequency channels (highest freq = last channels)
    n_top = max(1, int(n_freq * freq_frac))  # ~25 channels
    top_band = patch[:, -n_top:]  # (256, ~25)
    
    # Mean power per time bin across just those channels
    profile = np.mean(top_band, axis=1)  # (256,)
    
    # Robust background: median + MAD
    med = np.median(profile)
    mad = np.median(np.abs(profile - med))
    sigma = mad * 1.4826  # MAD → std conversion
    
    if sigma < 1e-10:
        # Flat profile, try raw std as fallback
        sigma = np.std(profile)
        if sigma < 1e-10:
            return None
    
    # First bin exceeding threshold
    above = np.where(profile > med + threshold_sigma * sigma)[0]
    if len(above) == 0:
        return None  # no detectable burst
    return above[0]


def find_burst_duration(patch, onset, threshold_sigma=2.5, freq_frac=0.1):
    """
    Total burst duration in bins: span from onset to the last bin still
    above the same top-band significance threshold used by find_burst_onset.
    No per-burst duration is stored in the dataset, so this is recomputed
    directly from the patch (raw sim) data rather than assumed.
    """
    if onset is None:
        return 1
    n_time, n_freq = patch.shape
    n_top = max(1, int(n_freq * freq_frac))
    top_band = patch[:, -n_top:]
    profile = np.mean(top_band, axis=1)
    med = np.median(profile)
    mad = np.median(np.abs(profile - med))
    sigma = mad * 1.4826
    if sigma < 1e-10:
        sigma = np.std(profile)
        if sigma < 1e-10:
            return 1
    above = np.where(profile > med + threshold_sigma * sigma)[0]
    above_after_onset = above[above >= onset]
    if len(above_after_onset) == 0:
        return 1
    return max(int(above_after_onset.max()) - int(onset) + 1, 1)


def measure_detection_latency(model, patch, n_steps=256, fill_value=None,
                               trigger_thresh=0.5, converge_thresh=0.95):
    """
    Run progressive reveal at single-bin resolution.
    Returns: (burst_onset_bin, trigger_bin, converge_bin, time_fracs, probs)
    
    trigger_bin and converge_bin are absolute positions; subtract burst_onset_bin
    to get latency in timesteps after burst.

    All masked variants for this patch are built up front and scored in a single
    batched model.predict() call instead of one predict() call per bin (same
    math/output, ~256x fewer call-overhead hits per patch).
    """
    n_time, n_freq = patch.shape
    step_size = max(1, n_time // n_steps)
    bins = np.arange(step_size, n_time + 1, step_size)

    batch = np.empty((len(bins), n_time, n_freq), dtype=patch.dtype)
    for i, t in enumerate(bins):
        masked = np.copy(patch)
        if fill_value is None:
            masked[t:, :] = masked[:t, :].mean()
        elif fill_value == 'noise':
            revealed = masked[:t, :]
            masked[t:, :] = np.random.normal(
                revealed.mean(), max(revealed.std(), 1e-6), masked[t:, :].shape
            )
        else:
            masked[t:, :] = fill_value
        batch[i] = masked

    probs = model.predict(batch[..., np.newaxis], verbose=0, batch_size=256).flatten()

    burst_onset = find_burst_onset(patch)
    
    # Find first bin where P exceeds thresholds
    trigger_bin = None
    converge_bin = None
    
    trigger_idxs = np.where(probs >= trigger_thresh)[0]
    if len(trigger_idxs) > 0:
        trigger_bin = bins[trigger_idxs[0]]
    
    converge_idxs = np.where(probs >= converge_thresh)[0]
    if len(converge_idxs) > 0:
        converge_bin = bins[converge_idxs[0]]
    
    return burst_onset, trigger_bin, converge_bin, bins, probs


# --- Run on ALL FRB test examples ---
frb_test_idxs = np.where(y_test == 1)[0]

results = []
for idx in frb_test_idxs:
    patch = X_test[idx, :, :, 0]
    onset, trig, conv, bins, probs = measure_detection_latency(
        model, patch, n_steps=256, fill_value=None
    )
    if onset is not None and trig is not None:
        results.append({
            'idx': idx,
            'burst_onset': onset,
            'trigger_bin': trig,
            'converge_bin': conv,
            'bins_to_trigger': trig - onset,
            'bins_to_converge': conv - onset if conv is not None else None,
            'bins': bins,
            'probs': probs,
            'duration': find_burst_duration(patch, onset),
        })

print(f"Measured {len(results)} / {len(frb_test_idxs)} FRB patches "
      f"(skipped {len(frb_test_idxs) - len(results)} with no detectable burst or no trigger)\n")

# --- Statistics ---
bins_to_trigger = np.array([r['bins_to_trigger'] for r in results])
bins_to_converge = np.array([r['bins_to_converge'] for r in results 
                              if r['bins_to_converge'] is not None])

EFFECTIVE_TIME_RES = 2.56e-6 * 256  # seconds per bin (~0.655 ms)

print("=" * 60)
print("BINS after burst onset to P > 0.5 (trigger):")
print(f"  Mean:   {np.mean(bins_to_trigger):.1f} bins  "
      f"({np.mean(bins_to_trigger) * EFFECTIVE_TIME_RES * 1e3:.2f} ms)")
print(f"  Median: {np.median(bins_to_trigger):.1f} bins  "
      f"({np.median(bins_to_trigger) * EFFECTIVE_TIME_RES * 1e3:.2f} ms)")
print(f"  Std:    {np.std(bins_to_trigger):.1f} bins")
print(f"  Min:    {np.min(bins_to_trigger)} bins  "
      f"({np.min(bins_to_trigger) * EFFECTIVE_TIME_RES * 1e3:.2f} ms)")
print(f"  Max:    {np.max(bins_to_trigger)} bins  "
      f"({np.max(bins_to_trigger) * EFFECTIVE_TIME_RES * 1e3:.2f} ms)")

print()
print("BINS after burst onset to P > 0.95 (convergence):")
print(f"  Mean:   {np.mean(bins_to_converge):.1f} bins  "
      f"({np.mean(bins_to_converge) * EFFECTIVE_TIME_RES * 1e3:.2f} ms)")
print(f"  Median: {np.median(bins_to_converge):.1f} bins  "
      f"({np.median(bins_to_converge) * EFFECTIVE_TIME_RES * 1e3:.2f} ms)")
print(f"  Std:    {np.std(bins_to_converge):.1f} bins")
print(f"  Min:    {np.min(bins_to_converge)} bins  "
      f"({np.min(bins_to_converge) * EFFECTIVE_TIME_RES * 1e3:.2f} ms)")
print(f"  Max:    {np.max(bins_to_converge)} bins  "
      f"({np.max(bins_to_converge) * EFFECTIVE_TIME_RES * 1e3:.2f} ms)")

# --- Plot ---
TITLE_FS, LABEL_FS, TICK_FS, LEGEND_FS = 18, 15, 13, 12
fig, axes = plt.subplots(1, 3, figsize=(22, 6))

# Panel 1: Histogram of bins to trigger
ax = axes[0]
ax.hist(bins_to_trigger, bins=30, edgecolor='black', alpha=0.7, color='steelblue')
ax.axvline(np.median(bins_to_trigger), color='red', ls='--', lw=2, 
           label=f'median = {np.median(bins_to_trigger):.0f} bins')
ax.set_xlabel('Time bins after burst onset', fontsize=LABEL_FS)
ax.set_ylabel('Count', fontsize=LABEL_FS)
ax.set_title('Bins to P(FRB) > 0.5', fontsize=TITLE_FS)
ax.tick_params(labelsize=TICK_FS)
ax.legend(fontsize=LEGEND_FS)
ax.grid(True, alpha=0.3)

# Panel 2: Histogram of bins to convergence
ax = axes[1]
ax.hist(bins_to_converge, bins=30, edgecolor='black', alpha=0.7, color='darkorange')
ax.axvline(np.median(bins_to_converge), color='red', ls='--', lw=2,
           label=f'median = {np.median(bins_to_converge):.0f} bins')
ax.set_xlabel('Time bins after burst onset', fontsize=LABEL_FS)
ax.set_ylabel('Count', fontsize=LABEL_FS)
ax.set_title('Bins to P(FRB) > 0.95', fontsize=TITLE_FS)
ax.tick_params(labelsize=TICK_FS)
ax.legend(fontsize=LEGEND_FS)
ax.grid(True, alpha=0.3)

# Panel 3: Probability curves aligned to burst onset, normalized to % of FRB lifetime
ax = axes[2]
show_idxs = rng.choice(len(results), min(10, len(results)), replace=False)
for i in show_idxs:
    r = results[i]
    # Shift bins so burst_onset = 0, then keep only the post-onset lifetime
    shifted = r['bins'] - r['burst_onset']
    mask = shifted >= 0
    pct_lifetime = shifted[mask] / r['duration'] * 100
    ax.plot(pct_lifetime, r['probs'][mask], alpha=0.75, linewidth=1.4,
            label=f"FRB #{r['idx']}")

ax.axhline(0.5, color='r', ls='--', alpha=0.5, label='P = 0.5')
ax.axhline(0.95, color='orange', ls='--', alpha=0.5, label='P = 0.95')
ax.set_xlim(left=0)
ax.set_xlabel('% of FRB lifetime', fontsize=LABEL_FS)
ax.set_ylabel('P(FRB)', fontsize=LABEL_FS)
ax.set_title('Probability curves aligned to burst onset', fontsize=TITLE_FS)
ax.tick_params(labelsize=TICK_FS)
ax.legend(fontsize=LEGEND_FS, ncol=2, loc='lower right')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('figures/detection_latency.svg')
plt.show()

# --- Also report as fraction of the 256-bin window ---
print(f"\nAs fraction of full window (256 bins):")
print(f"  Trigger:    {np.median(bins_to_trigger)/256:.1%} of window after burst onset")
print(f"  Converge:   {np.median(bins_to_converge)/256:.1%} of window after burst onset")
```
